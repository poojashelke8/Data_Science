```python
# pip install pandas
import pandas as pd

sales_data = pd.read_csv(
     r"C:\Users\Shree\Desktop\Data_Science\Data_practice\Libraries\Visualization\Matplotlib\Datasets\sales_record\1000 Sales Records.csv"
)

sales_data.head()
```

```python
# check with missing val,info etc
sales_data.info()
sales_data.count()
sales_data.describe()

# with above there is no empty data we can move further
```

```python
# check with total revenue = unit sold * unit 
# .isclose() method checks whether two values are close to each other, or not. Returns True if the values are close, otherwise False.
import numpy as np
sales_data['Expected_revnue'] = sales_data['Units Sold'] * sales_data['Unit Price']
sales_data['revenue_matched'] = np.isclose(
    sales_data['Total Revenue'],
    sales_data['Expected_revnue']
)

# checked with if any no mathced revnue data
mismath_revnue = sales_data[~sales_data['revenue_matched']]
mismath_revnue

# check with expected profit = total revnue - total cost

sales_data['expected_profit'] = sales_data['Total Revenue'] - sales_data['Total Cost']
sales_data['Profit_matched'] = np.isclose(
    sales_data['Total Profit'],
    sales_data['expected_profit']
)

mismath_profit = sales_data[~sales_data['Profit_matched']]
mismath_profit
```

```python
# data understanding and cleaning
# 1. covert date 
sales_data['Order Date'] = pd.to_datetime(sales_data['Order Date'])
sales_data['Ship Date'] = pd.to_datetime(sales_data['Ship Date'])

# 2. create useful features
sales_data['Delivary days'] = (sales_data['Ship Date']-sales_data['Order Date']).dt.days
sales_data['Month'] = (sales_data['Order Date']).dt.month
sales_data['Year'] = (sales_data['Order Date']).dt.year

# deleted the col
# sales_data.drop(columns=['Dilivary days'], inplace=True)
sales_data
```

```python
# Univariate Analysis
# individual numerical col analysis
# Typical values(mean.median),spread(variance,IQR),shape(skewed/normal),Outliers,Business implications

num_cols = [
    'Units Sold',
    'Unit Price',
    'Unit Cost',
    'Total Revenue',
    'Total Cost',
    'Total Profit'
]
sales_data[num_cols].head()

# describe() gives a statistical summary of numerical columns.
# mean > median → right skew
# mean < median → left skew
sales_data[num_cols].describe()


# Mean vs Median (50%)
# Large gap → skewed data
# Min / Max
# Check for negative or extreme values
# Std
# High std → high variability
# 📌 Example insight:
# If mean profit >> median profit → few orders generate most profit
```

```python
# Distribution Analysis (Histogram)
# To understand: Data skewness,Long tails,Natural ranges
import matplotlib.pyplot as plt

for col in num_cols:
    plt.figure()
    plt.hist(sales_data[col], bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# Interpret: Sales & profit data are usually right-skewed.
# Right-skewed → few large values
# Left-skewed → many high values, few low
# Symmetric → near normal
```

```python
# Boxplot - outliers detection
for col in num_cols:
    plt.figure()
    plt.boxplot(sales_data[col], vert=False)
    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)
    plt.show()
```

```python
# check shewness
sales_data[num_cols].skew()
# Skew value	Meaning
# ≈ 0	        Symmetric
# > 1	        Highly right skewed
# < -1	        Highly left skewed
```

```python
# detect ouliers
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return series[(series < lower) | (series > upper)]

# Example: Outliers in Total Profit
profit_outliers = detect_outliers_iqr(sales_data['Total Profit'])
profit_outliers.head()
sales_data[sales_data['Total Profit'] < 0][num_cols].head()
len(profit_outliers)
```

```python
# Insight:
# Total Profit shows strong right skewness, indicating that a small number of orders contribute disproportionately to overall profitability.
```

```python
# Bivariate Analysis [How one variable affects another]
# Bivariate answers:
# “WHY is it happening?”
# “WHAT influences WHAT?” -->  Without bivariate analysis, your EDA stays descriptive, not analytical.

# [rerequisite] - convert date and create profit margin
sales_data['Profit Margin'] = sales_data['Total Profit'] / sales_data['Total Revenue']
sales_data
```

```python
# 1) Unit sold vs Total Profit  --> [volume vs profitability], selling more == high profit , identify large volume and low profit
# Uisng Scatter plot
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.scatter(sales_data['Units Sold'], sales_data['Total Profit'], alpha=0.5)
plt.xlabel('Units Sold')
plt.ylabel('Total Profit')
plt.title('Units Sold vs Total Profit')
plt.show()
```

```python
# unit price vs unit sold --> Reveals price elasticity ,Shows whether customers prefer low-price high-volume items

plt.figure(figsize=(6,4))

plt.scatter(sales_data['Unit Price'],sales_data['Units Sold'], alpha=0.3)
plt.ylabel('Units Sold')
plt.xlabel('Unit price')
plt.title('Units Sold vs Unit price')
plt.show()
```

```python
# Item type vs Total profit
# using bar plot

item_profit = (
    sales_data.groupby('Item Type')['Total Profit']
    .sum()
    .sort_values(ascending=True)
)

item_profit
plt.figure(figsize=(6,4))
item_profit.plot(kind='bar')
plt.xlabel('Item Type')
plt.ylabel('Total Profit')
plt.title("Total profit by item type")
plt.show()

# Cosmetics and household contribute the highest total profit, making them key revenue drivers.
```

```python
# sales channel vs profit margin
plt.figure(figsize=(6,4))
# using pandas boxplot
sales_data.boxplot(column="Profit Margin",by="Sales Channel")

plt.title("Profit margin by sales channel")
plt.suptitle('')
plt.xlabel('Sales channel')
plt.ylabel('Profit Margin')
plt.show()
```

```python
# sales channel vs profit margin
plt.figure(figsize=(6,4))

# using purely matplot lib - manual way
online = sales_data[sales_data['Sales Channel'] == 'Online']['Profit Margin']
offline = sales_data[sales_data['Sales Channel'] == 'Offline']['Profit Margin']

plt.boxplot([online,offline],labels=['Online','Offline'])
plt.title("Profit margin by sales channel")
plt.suptitle('')
plt.xlabel('Sales channel')
plt.ylabel('Profit Margin')
plt.show()
```

```python
# Delivery Days vs Total Profit --> Operational impact on profit
# Combines operations + finance

plt.figure()
plt.scatter(sales_data['Delivary days'], sales_data['Total Profit'], alpha=0.6)
plt.xlabel('Delivery Days')
plt.ylabel('Total Profit')
plt.title('Delivery Days vs Total Profit')
plt.show()

# Orders with longer delivery times tend to have lower profitability, indicating potential logistics inefficiencies.
```
