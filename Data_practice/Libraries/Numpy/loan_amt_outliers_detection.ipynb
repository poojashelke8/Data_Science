```python
import numpy as np

data = np.genfromtxt(
    r"C:\Users\Shree\Desktop\Data_Science\Data_practice\Libraries\Numpy\Datasets\Outlier_Loan_dataset.csv",
    delimiter=",",
    skip_header=1,
    dtype=str
)
print(data[:5])
```

```python
# step 2 - check with missing/null values
miss_val = (data == "")
np.sum(miss_val,axis=0)

```

```python
# step 3 - convert the loan amount column into float as we iitially converted to str (dtype = str)
np.unique(data[:,8]) #check the empty or negative values if present in col

load_col_data = data[:,8].copy()
# replace empty values with NAN
load_col_data[load_col_data == ""] = np.nan
# then convert to float
loan_amount = load_col_data.astype(float)
print(loan_amount[:10])
```

```python
# step 4 - handle missing values [instead of nan replace it with median]

# meadian_val = np.nanmedian(loan_amount)
# print(meadian_val)
# # replace the median wherever the nan present
# loan_amount = np.where(np.isnan(loan_amount),meadian_val,loan_amount)
# print(loan_amount[:10])
valid_mask = loan_amount >= 0

loan_amount = loan_amount[valid_mask]
data = data[valid_mask]   # keep dataset aligned
median_val = np.nanmedian(loan_amount)
print("Median Loan Amount:", median_val)

loan_amount = np.where(
    np.isnan(loan_amount),
    median_val,
    loan_amount
)
```

```python
# step 5 - outliers detection (IQR) - q3-q1
q1 = np.percentile(loan_amount,25)
q3 = np.percentile(loan_amount,75)

IQR  = q3-q1
print(IQR)

lower = q1 - 1.5 * IQR 
upper = q3 + 1.5 * IQR

outliers = loan_amount[(loan_amount<lower) | (loan_amount >upper)]
print(outliers)
```

```python
# step 6 - capping the outliers [it replaces with lower and upper,no rows deleted]
loan_amt_capped = np.clip(loan_amount,lower,upper)
```

```python
# step 7 - remove outliers(masking) and rebuild clean dataset [rows will be removed ]


clean_mask = (loan_amount >= lower) & (loan_amount <= upper)
loan_amount_clean = loan_amount[clean_mask]

print(loan_amount_clean)
```
