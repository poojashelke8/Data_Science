```python
import numpy as np

np.random.seed(42)

# created random marks for 20 students in the range(50 - 100)
# 4 subjects - math,history,science,IT
scores = np.random.randint(50,101,size=(20,4))
print(scores)

# return the shape,dimension,size
print(scores.shape)
print(scores.size)
print(scores.ndim)
```

```python
# return top 5 and lats 5 result
top_5 = scores[:6]
last_5 = scores[-5:]
print("Scores for first 5",top_5)
print("Scores for last 5",last_5)

# retrieve marks for specific sub
math_marks = scores[:,0]
IT_marks = scores[:,-1]

avg_math = np.mean(math_marks)
avg_IT = np.mean(IT_marks)

# used axis 
# axis=0 → collapse rows → keep columns
# axis=1 → collapse columns → keep rows
avg_all = np.mean(scores,axis=0)
print("Math scores all",math_marks)
print("IT scores all",IT_marks)

print("Average of maths",avg_math)
print("Average of IT",avg_IT)
print("Average of all subjects",avg_all)
```

```python
# Total scores of each student 
total_scores = scores.sum(axis=1)
print(total_scores)

# using python
for ind,res in enumerate(total_scores,start=1):
    print(f"{ind}. {res}")

# using numpy
indexed_score = np.column_stack((np.arange(1,len(total_scores)+1),(total_scores)))
print(indexed_score)
```

```python
# Performance Comparison
math_vs_history = scores[:,0]>scores[:,1]
print(math_vs_history)

# count students
total_stud = np.count_nonzero(math_vs_history)
print("Students math marks > history marks -->",total_stud)
```

```python
# normalization
# It’s a performance profile, not just marks.

# column wise normalization
max_nor =scores.max(axis=0)
min_nor = scores.min(axis=0)
normalized_score = (scores-min_nor)/(max_nor-min_nor)
readable = np.round(normalized_score,2)
print(normalized_score)
print(normalized_score[:,0])
print(readable)


```

```python
# row wise normalization
row_sum = scores.sum(axis=1, keepdims=True)
row_normalized = scores / row_sum
print(row_normalized)
```

```python
# average for all subjects
subjects = ["Math", "History", "Science", "IT"]

for i, subj in enumerate(subjects):
    print(subj, "avg:", scores[:, i].mean())
```

```python
rank_normal = normalized_score.sum(axis=1)
print(rank_normal)

# np.argsort(array) returns the indices that would sort the array.
# It does not return the sorted values itself.
# It returns an array of positions (indices).

ranks = np.argsort(-rank_normal) 
print(ranks)

for rank,stud_ind in enumerate(np.argsort(ranks),start=1):
    print(f"{rank} student {stud_ind+1} Score {rank_normal[stud_ind]: .2f} ")


# top performer

top_performer = np.argmax(normalized_score,axis=0)
subjects = ["Math", "History", "Science", "IT"]
print("Top Performer Subject")
for subj,stud_ind in enumerate(top_performer):
    print(f"Subject {subjects[subj]} : Student {stud_ind+1}")
```
