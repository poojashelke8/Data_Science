```python
# Series - Dataset with 1-D shapes,likely an array or list
# can store any kind of object also can customize 
# its index with diff num,int,str 
import pandas as pd

sr = pd.Series([1,2,3,4,5])
sr.index=["p","o","o","j","a"]
sr
```

```python
# Customize index while creating a series

sr_index = pd.Series([40,50,60,70,80,90,100],
                     index = ['a','b','c','d','e','f','g'],
                     name="New Series")
sr_index
sr_index['f']
sr_index['e'] = 1
sr_index['b':'e']

# calling with index using
# 1) .loc
sr_index.loc['e':'f']

# 2) .iloc --> access object cells using a numeric index
sr_index.iloc[1:4]
```

```python
sr_index>=70
sr_index[sr_index>80]

# add/merge a series with another series
sr_new = pd.Series([33,55,88])

sr_new_series = pd.concat([sr_index, sr_new])
sr_new_series
```
