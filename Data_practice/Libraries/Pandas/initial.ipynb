```python
# missing Values
loan_data = pd.read_csv("Outlier_Loan_dataset.csv")
loan_data.head()

# loan_data.isna()
# loan_data.isnull()
loan_data.info()
loan_data.count()


```

```python
# can fill or clean data as per requirement
# .fillna(val) --> replaces all NAN val ith static value

loan_data.fillna(0)
# can fills missing value in a specific column
loan_data['Age'].fillna(10)
# loan_data.ffill()--> ffill()/bfill - fills missing value using prev or next row
```

```python
# Dropping missing values

# .dropna()--> removes rows that contain atleast 1 missing value
loan_data.dropna()
# dropnna(how='all') --> removes only rows where all values missing
loan_data.dropna(how='all')
# removes any column that containes at least one missing value
loan_data.dropna(axis=1)
# df.dropna(inplace=True): Modifies the original DataFrame directly rather than returning a new one.
```

```python
# Data type and conversion

loan_data['Loan_Duration_Months'] = loan_data['Loan_Duration_Months'].astype(float)
loan_data
```

```python
loan_amt = loan_data['Loan_Amount'].copy()
loan_amt = pd.to_numeric(loan_amt,errors='coerce')

# before cleaning
print("missing",loan_amt.isna().sum())
print(loan_amt.head())

# fill missing value with median
meadian_val = loan_amt.median()
loan_amt = loan_amt.fillna(meadian_val)

print("Median",meadian_val)
print("Missing after fillna",loan_amt.isna().sum())

# cleaning data
# i.e loan amount can never be negative
invalid_cnt = (loan_amt<=0).sum()
print("Invalid Values",invalid_cnt)
loan_amt[loan_amt<=0] = meadian_val

# Outlier handle
# IQR calculate
# In pandas to calculate percentile same like in numpy
# -- use quantile and val
q1 = loan_amt.quantile(0.25)
q3 = loan_amt.quantile(0.75)

IQR = q3-q1
lower = q1 -1.5 *IQR
upper = q3 + 1.5 * IQR

print("Q1 -->",q1,"Q3 -->",q3)
print("IQR -->",IQR)
print("upper Bound -->",upper,"lower Bound -->",lower)

# detect outliers
outliers = loan_amt[(loan_amt < lower) | (loan_amt > upper)]
print("outliers count",outliers.shape[0])

# replae outliers with median
loan_amt_clean = loan_amt.copy()
loan_amt_clean[loan_amt_clean < lower] = meadian_val
loan_amt_clean[loan_amt_clean > upper] = meadian_val

# After Cleaning
print("Missing values",loan_amt_clean.isna().sum())
print("sample Clean data",loan_amt_clean.head())

# Assigning back to main
loan_data['Loan_Amount_Cleaned'] = loan_amt_clean
```

```python
loan_data
```

```python
# If you want to create a new column → use transform()
# If you want a summary table → use agg()


# Groupby --> allowing for aggregation,transformation,filtering data
# core concept --> SPLIT - APPLY - COMBINE

loan_data.groupby('Job', as_index=False)['Loan_Amount_Cleaned'].mean()

# output : Output rows = number of jobs , You cannot directly assign it back to the original DataFrame

# multiple aggregation
loan_data.groupby('Job', as_index=False)['Loan_Amount_Cleaned'].agg(['mean','median','count'])
```

```python
# Transform --> transform() applies a function group-wise BUT returns an output with the SAME size as the original DataFrame.
loan_data['loan_avg_data'] = (
    loan_data.groupby('Job')['Loan_Amount_Cleaned']
    .transform('mean')
)
loan_data
# above returning new col with avg loan data for each job profession for individual row no entirely
# Each row gets its job’s average loan amount

# Normalize within group
# We divide each customer’s cleaned loan amount by the average loan amount of their job using groupby().transform('mean').
# This creates a normalized feature showing how large the loan is relative to peers with the same job.

import numpy as np


loan_data['Loan_vs_Job_Avg'] = (
    loan_data['Loan_Amount_Cleaned'] /
    loan_data.groupby('Job')['Loan_Amount_Cleaned'].transform('mean').replace(0,np.nan)
)

loan_data
```

```python
# Loan_vs_Job_Avg --> “How large is this person’s loan compared to others with the same job?”
loan_data['Loan_Risk_Level'] = pd.cut(
    loan_data['Loan_vs_Job_Avg'],
    bins=[0, 0.8, 1.2, np.inf],
    labels=['Low Risk', 'Medium Risk', 'High Risk']
)

loan_data
```
