```python
# pip install pandas
import pandas as pd

sales_data = pd.read_csv(
     r"C:\Users\Shree\Desktop\Data_Science\Data_practice\Libraries\Visualization\Matplotlib\Datasets\sales_record\1000 Sales Records.csv"
)

sales_data.head()
```

```python
# check with missing val,info etc
sales_data.info()
sales_data.count()
sales_data.describe()

# with above there is no empty data we can move further
```

```python
# check with total revenue = unit sold * unit 
# .isclose() method checks whether two values are close to each other, or not. Returns True if the values are close, otherwise False.
import numpy as np
sales_data['Expected_revnue'] = sales_data['Units Sold'] * sales_data['Unit Price']
sales_data['revenue_matched'] = np.isclose(
    sales_data['Total Revenue'],
    sales_data['Expected_revnue']
)

# checked with if any no mathced revnue data
mismath_revnue = sales_data[~sales_data['revenue_matched']]
mismath_revnue

# check with expected profit = total revnue - total cost

sales_data['expected_profit'] = sales_data['Total Revenue'] - sales_data['Total Cost']
sales_data['Profit_matched'] = np.isclose(
    sales_data['Total Profit'],
    sales_data['expected_profit']
)

mismath_profit = sales_data[~sales_data['Profit_matched']]
mismath_profit
```

```python
# data understanding and cleaning
# 1. covert date 
sales_data['Order Date'] = pd.to_datetime(sales_data['Order Date'])
sales_data['Ship Date'] = pd.to_datetime(sales_data['Ship Date'])

# 2. create useful features
sales_data['Delivary days'] = (sales_data['Ship Date']-sales_data['Order Date']).dt.days
sales_data['Month'] = (sales_data['Order Date']).dt.month
sales_data['Year'] = (sales_data['Order Date']).dt.year

# deleted the col
# sales_data.drop(columns=['Dilivary days'], inplace=True)
sales_data
```
