```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/titanic.csv")
df.head()
```

```python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head()
```

```python
target = df.Survived
inputs = df.drop('Survived',axis='columns')
```

```python
# manage sex feature 
dummy = pd.get_dummies(inputs['Sex']).astype(int)
dummy
```

```python
inputs = pd.concat([inputs,dummy],axis='columns')
inputs.head()
```

```python
inputs.drop(['Sex','male'],axis='columns',inplace=True)
inputs.head(3)
```

```python
# check if na value is there or not
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age[:10]
```

```python
# we have nan value in age, replacing it with mean of that column

inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head(10)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(inputs,target,test_size=0.2)
```

```python
# using Guassian Naive bayes - used when data distribution normal
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```
